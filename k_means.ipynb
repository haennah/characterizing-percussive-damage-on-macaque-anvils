```python
# K means cluster on the textured faces of an obj file.
# Sample texture colors at face centers as an average of the vertex values and cluster based on HSV values, not RGB.
# Then extract the masked faces based on selected k means clusters.

from cv2 import cvtColor, COLOR_BGR2HSV
import numpy as np
from sklearn.cluster import KMeans
import trimesh
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib import cm
import os

# Load mesh
path = "FILEPATH"  # specify your path here
filename = "FILENAME.obj"
obj_path = os.path.join(path, filename)

# Load the textured OBJ file using trimesh
print(f"Loading mesh {obj_path}...")
mesh = trimesh.load(obj_path, force='mesh') # load the mesh
vertices = mesh.vertices # get vertx info
faces = mesh.faces # get face info
print(f"Loaded mesh with {len(vertices)} vertices and {len(faces)} faces")
```

```python
# Check for UV coordinates and texture image
print("Checking for UV coordinates and texture image...")
if hasattr(mesh.visual, 'uv') and mesh.visual.uv is not None: # check if UV coordinates exist
    uv_coords = mesh.visual.uv
else:
    raise ValueError("No UV coordinates found")

if hasattr(mesh.visual, 'material') and hasattr(mesh.visual.material, 'image'): # check if texture image exists
    texture_image = mesh.visual.material.image
else:
    raise ValueError("No texture image found in material")

texture_np = np.array(texture_image.convert('RGB')) # convert texture image to numpy array in RGB format
print("Texture loaded")

# Sample face color at face center UVs
print("Sampling texture colors at face centers...")
face_uvs = uv_coords[faces] # get UV coordinates for each vertex of each face
face_uvs_avg = face_uvs.mean(axis=1) # average UV coordinates for each face from the 3 vertices to get a single UV coordinate per face

# Flip V axis of the UV map to match image coordinates in the numpy array. To avoid distorted projected texture on the 3D mesh.
face_uvs_avg[:, 1] = 1.0 - face_uvs_avg[:, 1] 

# Sample the texture image to extract the colour of each face in the mesh.
uv_px = np.round(face_uvs_avg * [texture_np.shape[1] - 1, texture_np.shape[0] - 1]).astype(int) # scale uv coordinates to pixel coordinates in the textured image.
uv_px[:, 0] = np.clip(uv_px[:, 0], 0, texture_np.shape[1] - 1) # ensure pixel indices are within image bounds
uv_px[:, 1] = np.clip(uv_px[:, 1], 0, texture_np.shape[0] - 1) # ensure pixel indices are within image bounds
face_colors_rgb = texture_np[uv_px[:, 1], uv_px[:, 0]] # sample the RGB values of each face for plotting and clustering
face_colors_bgr = face_colors_rgb[..., ::-1] # convert RGB to BGR for OpenCV
face_colors_hsv = cvtColor(face_colors_bgr.reshape(-1, 1, 3), COLOR_BGR2HSV).reshape(-1, 3) # convert to HSV for clustering

print("Done")
```

```python
# K-means clustering
print("Performing K-means clustering on face colors...")
k = 8 # adjust the number of k clusters as needed
channels = [1,2]  # Select which HSV channels to use: 0=Hue, 1=Saturation, 2=Value
kmeans = KMeans(n_clusters=k, random_state=42, n_init='auto') # k means with k clusters, fixed random state for reproducibility and n_init to avoid warnings
face_labels = kmeans.fit_predict(face_colors_hsv[:, channels].astype(np.float32))
print("K-Means Done")
```

```python
# Elbow plot to determine optimal number of clusters
print("Computing elbow plot...")

inertias = []
k_range = range(2, 11)

for k_test in k_range:
    kmeans_test = KMeans(n_clusters=k_test, random_state=42, n_init='auto')
    kmeans_test.fit(face_colors_hsv[:,channels].astype(np.float32))
    inertias.append(kmeans_test.inertia_)
    print(f"k={k_test}: inertia={kmeans_test.inertia_:.2f}")

# Plot the elbow curve
plt.figure(figsize=(10, 6))
plt.plot(k_range, inertias, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Number of Clusters (k)', fontsize=12)
plt.ylabel('Inertia (Within-cluster sum of squares)', fontsize=12)
plt.title('Elbow Plot for K-Means Clustering', fontsize=14)
plt.grid(True, alpha=0.3)
plt.xticks(k_range)

os.makedirs('results', exist_ok=True)
plt.savefig('results/elbow_plot.png', dpi=300, bbox_inches='tight')
plt.show()

print("Elbow plot saved to results/elbow_plot.png")
```

```python
# Function to set equal aspect ratio for 3D plots to avoid distortion of plotted meshes on the 3d plot grid.
def set_axes_equal(ax, vertices):
    x_limits = [vertices[:, 0].min(), vertices[:, 0].max()]
    y_limits = [vertices[:, 1].min(), vertices[:, 1].max()]
    z_limits = [vertices[:, 2].min(), vertices[:, 2].max()]
    
    max_range = max(
        x_limits[1] - x_limits[0],
        y_limits[1] - y_limits[0],
        z_limits[1] - z_limits[0]
    ) / 2.0
    
    mid_x, mid_y, mid_z = np.mean(x_limits), np.mean(y_limits), np.mean(z_limits)
    
    ax.set_xlim(mid_x - max_range, mid_x + max_range)
    ax.set_ylim(mid_y - max_range * 1.2, mid_y + max_range * 1.2)
    ax.set_zlim(mid_z - max_range, mid_z + max_range)
```

```python
# I moved this to here so that I could use it for plotly.
cmap = cm.get_cmap('tab20', k)   # 'tab20' has up to 20 distinct colors, adjust if k > 20
colors = [cmap(i) for i in range(k)] # get the colors from the colormap
```

```python
# Plot clusters on mesh
print("Plotting clusters on mesh...")

# Create a downsample of face_centers and corresponding labels/colors for faster plotting
face_centers = mesh.triangles_center # get the center of each face for plotting
sample_size_centers = int(len(face_centers) * 0.7) # change value here to define downsample size
sample_indices_centers = np.random.choice(len(face_centers), size=sample_size_centers, replace=False)
face_centers_sample = face_centers[sample_indices_centers]
face_colors_rgb_sample = face_colors_rgb[sample_indices_centers]
face_labels_sample_centers = face_labels[sample_indices_centers]

# First subplot
fig = plt.figure(figsize=(20, 10))
ax = fig.add_subplot(121, projection='3d') # 1 row, 1 colunm, first subplot at position 1 with a 3d projection style
ax.set_title("Original Textured Mesh")
ax.scatter(face_centers_sample[:, 0], face_centers_sample[:, 1], face_centers_sample[:, 2], c=face_colors_rgb_sample / 255.0, s=0.2) # generate a scatterplot of the face centers colored by the sampled texture color
set_axes_equal(ax, vertices) # set equal aspect ratio so mesh isn't distorted using the function defined above
ax.view_init(elev=90, azim=0) # view plot from above by setting the elevation angle to 90

# Second subplot
ax2 = fig.add_subplot(122, projection='3d') # 1 row, 1 colunm, second subplot at position 2 with a 3d projection style
ax2.set_title("Face Cluster Mapping")
sc2 = ax2.scatter(
    face_centers_sample[:, 0], face_centers_sample[:, 1], face_centers_sample[:, 2],
    c=face_labels_sample_centers, cmap=cmap, s=0.2
) # scatterplot of the face centers colored by the k means cluster index

set_axes_equal(ax2, vertices) # set equal aspect ratio so mesh isn't distorted using the function defined above
ax2.view_init(elev=90, azim=0) # view plot from above by setting the elevation angle to 90

# Legend setup
counts = np.bincount(face_labels_sample_centers, minlength=k) # count the number of faces in each cluster
handles = [
    Line2D([0], [0], # X and Y set to 0 to make a single point
           marker='o', linestyle='', markersize=8, # marker style
           markerfacecolor=colors[i], markeredgecolor='k', # black edge for better visibility
           label=f'Cluster {i} (n={counts[i]})')
    for i in range(k)
]
ax2.legend(handles=handles, title="Clusters",
           loc='upper left', bbox_to_anchor=(1.02, 1), frameon=True) # place legend outside the plot

os.makedirs('results', exist_ok=True)
plt.savefig('results/side_by_side_kmeans.png', dpi=300, bbox_inches='tight')
plt.tight_layout() 
plt.show()
```

```python
# Calculate average saturation and brightness value for each cluster
import pandas as pd

print("\nAverage Saturation and Brightness for each cluster:")

brightness_saturation_stats = []

for i in range(k):
    cluster_mask = (face_labels == i)
    avg_saturation = np.mean(face_colors_hsv[cluster_mask, 1])
    avg_brightness = np.mean(face_colors_hsv[cluster_mask, 2])
    brightness_saturation_stats.append((i, avg_saturation, avg_brightness))
    print(f"Cluster {i}: Avg Saturation = {avg_saturation:.2f}, Avg Brightness = {avg_brightness:.2f}")
    
summary = pd.DataFrame.from_records(brightness_saturation_stats, columns=['Cluster', 'Avg_Saturation', 'Avg_Brightness'])

# Export to CSV
os.makedirs('results', exist_ok=True)
summary.to_csv('results/cluster_brightness_saturation_stats.csv', index=False, sep=';', decimal=',')
```

```python
# Plot average saturation and brightness as a cross plot
plt.figure(figsize=(8, 6))
plt.scatter(summary['Avg_Saturation'], summary['Avg_Brightness'], c=summary['Cluster'], cmap=cmap, s=100)

for i, row in summary.iterrows():
    plt.text(row['Avg_Saturation'], row['Avg_Brightness'], str(int(row['Cluster'])), fontsize=9, ha='center', va='center', color='white')

plt.xlabel('Average Saturation', fontsize=12)
plt.ylabel('Average Brightness', fontsize=12)
plt.title(f'Average Saturation vs Brightness KBN-AV6', fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()

# Save the saturation vs brightness plot
os.makedirs('results', exist_ok=True)
plt.savefig('results/avg_saturation_brightness.png', dpi=300)
```

```python
# Select damage clusters manually
# Covert topologic rings to approxiamte metric distances using average traingle edge length.

print("\nCluster stats:")
for i in range(k): # loop over each cluster
    count = np.sum(face_labels == i) # count the number of faces in this cluster
    print(f"Cluster {i}: {count} faces")

damage_clusters = [6]  # update based on visual inspection
print(f"\nSelected damage clusters: {damage_clusters}")

# Create adjacency buffer around selected damage clusters
mesh_full = trimesh.Trimesh(vertices=vertices, faces=faces, process=False)  # full mesh
adj = mesh_full.face_adjacency  # array of adjacent face pairs
n_faces = len(faces)  # number of faces

# Calculate average edge length and convert to metric buffer
edge_lengths = mesh_full.edges_unique_length  # in meters (mesh units)
avg_edge_length = np.mean(edge_lengths)  # in meters
buffer_distance_mm = 0.5 # Define your desired buffer distance in mm
buffer_distance_m = buffer_distance_mm / 1000.0  # Convert mm to meters
n_rings = int(np.ceil(buffer_distance_m / avg_edge_length)) # define number of rings to grow based on desired metric buffer distance and average edge length
print(f"\nBuffer calculation:")
print(f"  Average edge length: {avg_edge_length*1000:.3f} mm")
print(f"  Target buffer distance: {buffer_distance_mm} mm")
print(f"  Calculated n_rings: {n_rings}")

# faces in selected damage clusters
damage_mask = np.isin(face_labels, damage_clusters)  # create a boolean mask for faces in the damage clusters
damage_with_buffer_mask = damage_mask.copy()  # create a mask that will hold all faces in damage clusters plus buffer. a copy that will later hold the buffer expansion.

# frontier for BFS (current ring)
frontier_mask = damage_mask.copy() # hold the current ring of faces to expand from

for _ in range(n_rings):  # grow buffer ring by ring
    mask0 = frontier_mask[adj[:, 0]]  # select edges where face 0 is in frontier
    mask1 = frontier_mask[adj[:, 1]]  # select edges where face 1 is in frontier
    neighbors0 = adj[mask0, 1]  # where face 0 is in the frontier, get face 1 as neighbor
    neighbors1 = adj[mask1, 0]  # where face 1 is in the frontier, get face 0 as neighbor
    neighbors = np.concatenate([neighbors0, neighbors1])  # all neighbor faces
    neighbors = np.unique(neighbors)  # unique neighbor indices

    # keep only faces not already marked as damage/buffer
    neighbors = neighbors[~damage_with_buffer_mask[neighbors]]

    if neighbors.size == 0:  # stop if nothing new to add
        break

    damage_with_buffer_mask[neighbors] = True  # add new ring to damage+buffer

    frontier_mask = np.zeros(n_faces, dtype=bool)  # new frontier = this ring
    frontier_mask[neighbors] = True

# Boolean mask for faces NOT in the selected damage clusters (with buffer)
undamaged_faces_mask = ~damage_with_buffer_mask  # invert final damage+buffer mask
undamaged_faces = faces[undamaged_faces_mask] # extract the undamaged faces

# Remap vertices and faces for clean export
unique_vertices, inverse_indices = np.unique(undamaged_faces.reshape(-1), return_inverse=True) # get unique vertex indices and remap face indices
undamaged_vertices = vertices[unique_vertices] # extract the undamaged vertices
undamaged_faces_remapped = inverse_indices.reshape(-1, 3) # remap the face indices to the new vertex indices. -1 infers the number of faces automatically and 3 is for the 3 vertices per face

# Remap face colors to match the undamaged faces
undamaged_face_colors = face_colors_rgb[undamaged_faces_mask]

# Create new mesh without the damage that was selected in the damage clusters line
undamaged_mesh = trimesh.Trimesh(
    vertices=undamaged_vertices, 
    faces=undamaged_faces_remapped, 
    face_colors=undamaged_face_colors,
    process=False
)

# Create mesh containing only the selected damage clusters (core + N-ring buffer)
damaged_faces_mask = damage_with_buffer_mask  # use adjacency buffer mask
damaged_faces = faces[damaged_faces_mask]

# Remap vertices and faces for clean export
unique_vertices_damaged, inverse_indices_damaged = np.unique(damaged_faces.reshape(-1), return_inverse=True)
damaged_vertices = vertices[unique_vertices_damaged]
damaged_faces_remapped = inverse_indices_damaged.reshape(-1, 3)

# Remap face colors to match the damaged faces
damaged_face_colors = face_colors_rgb[damaged_faces_mask]

# Create new mesh with only the damage clusters
damaged_mesh = trimesh.Trimesh(
    vertices=damaged_vertices,
    faces=damaged_faces_remapped,
    face_colors=damaged_face_colors,
    process=False
)

# Export
damaged_path = obj_path.replace('.obj', f'_damaged_cluster6_buffered.obj')
damaged_mesh.export(damaged_path)
print(f"Damaged mesh exported to: {damaged_path}")
```
